# Inventory Optimization for NovaCart – Business Understanding

**Company:** NovaCart Online Retail Pvt. Ltd.  
**Fulfilment Centre:** East Zone Fulfilment Centre (EZFC)

This notebook sets the business context and defines the problem this project is trying to solve.
No modeling is done here. The focus is on understanding *why* the problem exists and *what success looks like*.

## Business Context

NovaCart operates a large pan-India e-commerce fulfilment network, serving millions of customers
across categories such as electronics, personal care, home furnishing, and fashion accessories.

Customer promises like:
- Next-Day Delivery
- Same-Day Dispatch
- Guaranteed Delivery Windows

are highly dependent on having the **right inventory available at the right warehouse at the right time**.

Over the past two quarters, the East Zone Fulfilment Centre (EZFC) has shown increasing inefficiencies
in inventory planning and replenishment.


## Key Business Problems Identified

1. **Stockouts of high-velocity SKUs**
   - Stockout rate has increased to ~7.8%
   - Leads to lost sales, cancellations, and poor customer experience

2. **Overstocking of long-tail SKUs**
   - Carrying costs increased by ~22%
   - Excess working capital locked in slow-moving inventory

3. **Inconsistent forecasting accuracy**
   - Works reasonably for some categories
   - Fails badly for seasonal and promotion-heavy products

4. **Unpredictable supplier lead times**
   - Ranges from 2 days (local vendors) to 18 days (imports)
   - Makes replenishment planning risky


## Why Simple Demand Forecasting Is Not Enough

Accurate demand forecasts alone do not guarantee good inventory decisions.

Inventory outcomes are also affected by:
- Forecast uncertainty
- Demand volatility
- Supplier lead time variability
- Desired service level targets

This project therefore focuses on **operationalizing forecasts** into:
- Safety stock levels
- Reorder points
- SKU-level replenishment strategies


## Business Objectives

The objective is to build a data-driven inventory replenishment framework that can be
integrated into NovaCart’s daily planning workflow.

The system should:
- Forecast SKU-level weekly demand
- Identify volatile and unpredictable SKUs
- Recommend optimal reorder points (ROP)
- Calculate safety stock using forecast error and lead time
- Segment SKUs into replenishment strategies (ABC/XYZ)


## Success Metrics

The project will be considered successful if it delivers measurable improvements in:

- **Service Level:** ≥ 96%
- **Stockout Rate:** ≤ 2%
- **Excess Inventory Reduction:** ≥ 15%
- **Improved predictability** for high-velocity and volatile SKUs

These metrics align with the priorities of:
- Supply Chain Planning
- Warehouse Operations
- Finance & Working Capital Teams


## Key Stakeholders

- **Supply Chain Planning Team**
  - Demand forecasting and replenishment cycles

- **Warehouse Operations Managers**
  - Storage utilization and pick efficiency

- **Finance & Working Capital Team**
  - Inventory carrying cost and cash flow

- **Category Managers**
  - Promotions and product lifecycle decisions

- **Vendor Management Teams**
  - Supplier lead time performance


## Project Roadmap

The project will be executed in the following stages:

1. Data validation and cleaning
2. Exploratory data analysis (EDA)
3. Feature engineering
4. SKU segmentation (ABC/XYZ)
5. Demand forecasting
6. Inventory optimization (ROP & safety stock)
7. Scenario simulation and stress testing
8. Business impact evaluation

Each stage builds logically on the previous one.
